```python
import pandas as pd
import sqlite3

# Loading the data
df = pd.read_csv("../data/TMDB_movie_dataset_v11.csv")

#let us see what we got!
df.head(10)
```

```python
#lets begin by taking what we think its important
movies = df[[
    "id",
    "title",
    "release_date",
    "genres",
    "original_language",
    "popularity",
    "vote_average",
    "vote_count",
    "budget",
    "revenue",
    "runtime",
    "spoken_languages"
]].copy()

movies.head()
```

```python
#and now the database itself
conn = sqlite3.connect("movies.db")
movies.to_sql("movies", conn, if_exists="replace", index=False)
```

```python
#lets begin querying :D
#i want to see the Top 10 highest rated movies (min 500 votes)

query = """
        SELECT title, popularity, vote_average
        from movies
        where vote_count > 500
        order by vote_average DESC
        LIMIT 10;
"""
df = pd.read_sql(query,conn)
display(df)
```

```python
#another one! 
#I want to see how many different language exist in the db
query = """
            SELECT title, original_language,popularity, count(original_language)
            from movies
            group by original_language
            order by popularity DESC


"""

df = pd.read_sql(query,conn)
display(df)
```

```python
#top 10 popular movies?
#don't search 3rd most popular movie...!
query ="""
    SELECT title, popularity, original_language, genres
    from Movies
    order by popularity DESC
    LIMIT 10

"""
df = pd.read_sql(query,conn)
display(df)
```

```python
#movies with highest budget
query = """

        SELECT title, budget
        from movies
        order by budget DESC
        limit 10


"""
df = pd.read_sql(query,conn)
display(df)
```

```python
#movies with the highest revenue
query = """

        SELECT title, revenue
        from movies
        order by revenue DESC
        limit 10


"""
df = pd.read_sql(query,conn)
display(df)
```

```python
#movies longer than 3 hours
query = """


    SELECT title, runtime
    FROM movies
    WHERE runtime > 180
    ORDER BY runtime DESC;

    
"""
df = pd.read_sql(query,conn)
display(df)
```

```python
#Average rating per language

query = """

            SELECT  original_language, round(vote_average) as average_rating_per_language
            from movies
            WHERE vote_average BETWEEN 1 AND 9.9
            group by original_language
            order by average_rating_per_language DESC
           
            

"""
df = pd.read_sql(query,conn)
display(df)
```

```python
#Average runtime per language

query = """
            SELECT original_language, round(runtime) as average_runtime_per_language
            from movies
            group by original_language
            order by average_runtime_per_language desc
            limit 10

"""


df = pd.read_sql(query,conn)
display(df)
```

```python
#Average rating in greek movies
query = """
            SELECT ROUND(AVG(vote_average), 2) AS avg_greek_rating
            FROM movies
            WHERE original_language = 'el' AND vote_average BETWEEN 1 AND 9.9


"""


df = pd.read_sql(query,conn)
display(df)
```

```python
#Average rating in english spoken movies


query = """
            SELECT ROUND(AVG(vote_average), 2) AS avg_en_rating
            FROM movies
            WHERE spoken_languages = 'English' AND vote_average BETWEEN 1 AND 9.9


"""


df = pd.read_sql(query,conn)
display(df)
```

```python
#Query 10: Movies released per day
query = """
SELECT release_date, COUNT(*) AS movie_count
        FROM movies
        WHERE release_date IS NOT NULL AND release_date < '2026-01-01'
        GROUP BY release_date
        ORDER BY release_date DESC
        """

df = pd.read_sql(query,conn)
display(df)

```

```python
#Movies per year

query = """
SELECT SUBSTR(release_date,1,4)as year, COUNT(*) AS movie_count
    FROM movies
    WHERE SUBSTR(release_date,1,4) IS NOT NULL AND release_date < 2027
    GROUP BY year
    ORDER BY year DESC
 

"""

df = pd.read_sql(query,conn)
display(df)


```

```python
#Most common genres
query = """
            SELECT genres, COUNT(*) AS count
            FROM movies
            GROUP BY genres
            ORDER BY count DESC
            LIMIT 10;
"""


df = pd.read_sql(query,conn)
display(df)
```

```python

#50 Movies with very high rating but low popularity (hidden gems)
query = """
        
        SELECT title, vote_average, popularity
        FROM movies
        WHERE vote_average BETWEEN 1 AND 9.9 AND popularity < 10
        ORDER BY vote_average DESC
        LIMIT 50

"""
df = pd.read_sql(query,conn)
display(df)
```

```python
# Average revenue by language

query = """
        SELECT original_language, ROUND(AVG(revenue), 0)/1000000 AS avg_revenue_in_M
        FROM movies
        WHERE revenue > 0
        GROUP BY original_language
        ORDER BY avg_revenue_in_M DESC

        """

df = pd.read_sql(query,conn)
display(df)
```

```python
# Movies with best revenue-to-budget ratio

query = """
        
SELECT title, budget, revenue,
       ROUND(1.0 * revenue / budget, 2) AS roi
FROM movies
WHERE budget > 100000 AND revenue > 0
ORDER BY roi DESC
LIMIT 10;

"""

df = pd.read_sql(query,conn)
display(df)
```

```python
#Biggest Box Office flops

query = """
    SELECT title, budget, revenue, (revenue - budget) AS loss
    FROM movies
    WHERE budget > 0 AND revenue > 0
    ORDER BY loss ASC
    LIMIT 10;

"""

df = pd.read_sql(query,conn)
display(df)
```
